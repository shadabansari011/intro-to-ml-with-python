# Neural Network on Breast Cancer


```python
from sklearn.datasets import load_breast_cancer
X= load_breast_cancer()
```

```python
X_train, X_test, y_train, y_test = train_test_split(X.data, X.target, stratify=X.target, random_state=0)
```

```python
from sklearn.neural_network import MLPClassifier
mlp = MLPClassifier(random_state=42)
mlp.fit(X_train, y_train)
```

```python
print("Accuracy on Train set :- {:.2f}".format(mlp.score(X_train, y_train)))
print("Accuracy on Test set :- {:.2f}".format(mlp.score(X_test, y_test)))
```

```python
plt.figure(figsize=(20, 5))
plt.imshow(mlp.coefs_[0], interpolation='none', cmap='viridis')
plt.yticks(range(30), X.feature_names)
plt.xlabel("Columns in weight matrix")
plt.ylabel("Input feature")
plt.colorbar()
```
