**1. Linear Regression (aka Ordinary least Square)**

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
import mglearn
```

```python
X, y = mglearn.datasets.make_wave(n_samples=60)
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42)
```

```python
lr = LinearRegression()
lr.fit(X_train, y_train)
```

```python
print("lr coef_ {}".format(lr.coef_))
print("lr intercept_ {}".format(lr.intercept_))
```

```python
print("Training Accuracy Score: {:.2f}".format(lr.score(X_train, y_train)))
print("Test Accuracy Score: {:.2f}".format(lr.score(X_test, y_test)))
```

```python
X, y = mglearn.datasets.load_extended_boston()
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=0)
```

```python
lr = LinearRegression()
lr.fit(X_train, y_train)
```

```python
print("Training Accuracy Score: {:.2f}".format(lr.score(X_train, y_train)))
print("Test Accuracy Score: {:.2f}".format(lr.score(X_test, y_test)))
```

**2. Ridge Regression**

```python
from sklearn.linear_model import Ridge
ridge = Ridge()
ridge.fit(X_train, y_train)
```

```python
print("Training Accuracy Score: {:.2f}".format(ridge.score(X_train, y_train)))
print("Testing Accuracy Score: {:.2f}".format(ridge.score(X_test, y_test)))
```

```python
ridge10 = Ridge(alpha=10)
ridge10.fit(X_train, y_train)
```

```python
print("Training Accuracy Score: {:.2f}".format(ridge10.score(X_train, y_train)))
print("Testing Accuracy Score: {:.2f}".format(ridge10.score(X_test, y_test)))
```

```python
ridge01 = Ridge(alpha=0.1)
ridge01.fit(X_train, y_train)
```

```python
print("Training Accuracy Score: {:.2f}".format(ridge01.score(X_train, y_train)))
print("Testing Accuracy Score: {:.2f}".format(ridge01.score(X_test, y_test)))
```

```python
import matplotlib.pyplot as plt
```

```python
plt.plot(ridge.coef_, 's', label="Ridge 1")
plt.plot(ridge10.coef_, '^', label="Ridge 10")
plt.plot(ridge01.coef_, 'v', label="Ridge 01")

plt.plot(lr.coef_, 'o', label="Linear Regression")
plt.xlabel("Coefficenet index")
plt.ylabel("Coefficenet magnitude")
plt.hlines(0, 0, len(lr.coef_))
plt.ylim(-25, 25)
plt.legend()
```

**3. Lasso Regression**

```python
from sklearn.linear_model import Lasso
X, y = mglearn.datasets.load_extended_boston()
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=0)
lasso = Lasso()
lasso.fit(X_train, y_train)
```

```python
print("Training Accuracy Score: {:.2f}".format(lasso.score(X_train, y_train)))
print("Testing Accuracy Score: {:.2f}".format(lasso.score(X_test, y_test)))
print("No. of features used: {}".format(np.sum(lasso.coef_ != 0)))
```

```python
lasso001 = Lasso(alpha=0.01, max_iter=100000)
lasso001.fit(X_train, y_train)
```

```python
print("Training set score: {:.2f}".format(lasso001.score(X_train, y_train)))
print("Test set score: {:.2f}".format(lasso001.score(X_test, y_test)))
print("No of features used: {}".format(np.sum(lasso001.coef_ != 0)))
```

```python
lasso00001 = Lasso(alpha=0.0001, max_iter=100000)
lasso00001.fit(X_train, y_train)
```

```python
print("Training set score: {:.2f}".format(lasso00001.score(X_train, y_train)))
print("Test set score: {:.2f}".format(lasso00001.score(X_test, y_test)))
print("Number of features used: {}".format(np.sum(lasso00001.coef_ != 0)))
```

```python
plt.plot(lasso.coef_, 's', label="Lasso alpha 1")
plt.plot(lasso001.coef_, '^', label="Lasso alpha 0.01")
plt.plot(lasso00001.coef_, 'v', label="Lasso alpha 0.001")

plt.plot(ridge01.coef_, 'o', label="Ridge alpha=0.1")
plt.legend(ncol=2, loc=(0, 1.05))
plt.ylim(-25, 25)
plt.xlabel("Coefficient index")
plt.ylabel("Coefficient magnitude")
```
