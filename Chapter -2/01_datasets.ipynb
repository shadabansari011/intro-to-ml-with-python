# Some Sample Datasets

1. Forge Datasets for Classification

```python
import mglearn
import matplotlib.pyplot as plt
import numpy as np
```

```python
X, y = mglearn.datasets.make_forge()
mglearn.discrete_scatter(X[:, 0],X[:, 1], y)
plt.legend(["Class 0", "Class 1"], loc=4)
plt.xlabel("First Feature")
plt.ylabel("Second Feature")
print("X.shape  {}".format(X.shape))
```

2. Wave Datasets for Regression

```python
X, y = mglearn.datasets.make_wave(n_samples=40)
plt.plot(X, y, 'o')
plt.ylim(-3, 3)
plt.xlabel("Feature")
plt.ylabel("Target")
```

3. Breast Cancer Datasets in sklearn for Classification

```python
from sklearn.datasets import load_breast_cancer
```

```python
cancer = load_breast_cancer()
print("Cancer Keys {}".format(cancer.keys()))
```

```python
print("Cancer Shape {}".format(cancer.data.shape))
```

```python
print("Sample count per class: \n {}".format({n:v for n,v in zip(cancer.target_names, np.bincount(cancer.target))}))
```

```python
print("Feature Names: {}".format(cancer.feature_names))
```
